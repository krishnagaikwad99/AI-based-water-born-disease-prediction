```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/dataset.csv")

print("Shape:", df.shape)

df.head()
```

```python
df["date"] = pd.to_datetime(df["date"])
print(df["date"].dtype)
```

```python
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
df = df.sort_values(by=["village_id", "date"])
df = df.reset_index(drop=True)
df.head(10)
```

```python
df[["ph_level", "turbidity_ntu", "residual_chlorine_mg_l", "rainfall_mm", "case_count"]].describe()
```

```python
import os
os.makedirs("data/processed", exist_ok=True)

df.to_csv("data/processed/cleaned_disease_dataset.csv", index=False)
print("Saved. Final shape:", df.shape)
```
